```python
# === 0) Librerías ===
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
```

```python
# === 1) Cargar tu dataset de características ===
df = pd.read_csv("caracteristicas.csv", parse_dates=["marca_tiempo"])
```

```python
# === 2) X, y (solo características; sin id/tiempo/estación) ===
y = df["hipoxia"].astype(int).values
Xdf = (df.drop(columns=["hipoxia","estacion","marca_tiempo"])
         .apply(pd.to_numeric, errors="coerce"))        # fuerza numérico; no convertibles -> NaN
# Imputación mínima para que ejecute siempre
Xdf = Xdf.groupby(df["estacion"]).ffill().bfill()       # por estación: relleno hacia adelante/atrás
Xdf = Xdf.fillna(Xdf.median(numeric_only=True))         # restante: mediana por columna
X = Xdf.values
cols = Xdf.columns.tolist()                             # orden de columnas (para nuevos datos)
print("Shapes -> X:", X.shape, " y:", y.shape)
```

```python
# === 3) Modelo RandomForest (baseline, sin tuning) ===
rf = RandomForestClassifier(random_state=42, n_jobs=-1) # configuración simple
rf.fit(X, y)                                            # ENTRENAR con TODO el dataset (como el profe)
print(rf)                                               # muestra el modelo
```

```python
# === 4) Predicción sobre datos nuevos (ejemplo) ===
# Opción A: usar 2 filas reales de tu dataset como “nuevas”
X_nuevo = X[:2]                                         # reemplaza por tus mediciones
pred = rf.predict(X_nuevo)
pred_proba = rf.predict_proba(X_nuevo)

print("Predicciones para datos nuevos:", pred)          # clases predichas (0=no hipoxia, 1=hipoxia)
print("Probabilidades para datos nuevos:\n", pred_proba)
```
