```python
# Modelo de aprendizaje profundo (Perceptrón Multicapa) para predecir hipoxia
# ============ 0) IMPORTS ============
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
# 1) Cargar dataset
datos = pd.read_csv("caracteristicas.csv")
X = datos.drop(columns=["hipoxia", "marca_tiempo", "estacion"], errors="ignore")
y = datos["hipoxia"].astype(int)
```

```python
# 2) Dividir en entrenamiento y prueba (70% - 30%)
X_ent, X_pru, y_ent, y_pru = train_test_split(
    X, y, test_size=0.3, stratify=y, random_state=42
)
```

```python
# 3) Escalar los datos (importante para redes neuronales)
escalador = StandardScaler()
X_ent = escalador.fit_transform(X_ent)
X_pru = escalador.transform(X_pru)
```

```python
# 4) Definir el modelo MLP
modelo = Sequential([
    Dense(64, activation="relu", input_shape=(X_ent.shape[1],)),  # capa oculta 1
    Dense(32, activation="relu"),                                # capa oculta 2
    Dense(1, activation="sigmoid")                               # salida (probabilidad de hipoxia)
])
```

```python
# 5) Compilar el modelo
modelo.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
# 6) Entrenar el modelo
historial = modelo.fit(
    X_ent, y_ent,
    epochs=10,             # entrenar 10 pasadas por los datos
    batch_size=32,         # tamaño de lote
    validation_data=(X_pru, y_pru),
    verbose=1
)
```

```python
# 7) Evaluación en el conjunto de prueba
loss, acc = modelo.evaluate(X_pru, y_pru, verbose=0)
print(f"\nExactitud en prueba: {acc:.3f}")
```
